```python
from pyspark.sql.functions import *

path = "/Volumes/workspace/default/mis_datasets_mhealth/"

columns = [
    "chest_acc_x",
    "chest_acc_y",
    "chest_acc_z",
    "ecg_lead_1",
    "ecg_lead_2",
    
    "ankle_acc_x",
    "ankle_acc_y",
    "ankle_acc_z",
    
    "ankle_gyro_x",
    "ankle_gyro_y",
    "ankle_gyro_z",
    
    "ankle_mag_x",
    "ankle_mag_y",
    "ankle_mag_z",
    
    "arm_acc_x",
    "arm_acc_y",
    "arm_acc_z",
    
    "arm_gyro_x",
    "arm_gyro_y",
    "arm_gyro_z",
    
    "arm_mag_x",
    "arm_mag_y",
    "arm_mag_z",
    
    "activity_label"
]

df = (
    spark.read
    .option("inferSchema", True)
    .option("delimiter", "\t")
    .csv(path)
)

df = df.toDF(*columns)

display(df)
```

```python
df.printSchema()
```

```python
from pyspark.sql.functions import *

df_bronze = (
    df
    .withColumn("source_file", col("_metadata.file_path"))
    .withColumn("ingestion_timestamp", current_timestamp())
)
```

```python
df_bronze = df_bronze.withColumn(
    "subject_id",
    regexp_extract(col("source_file"), r"subject(\d+)", 1)
)
```

```python
display(df_bronze)
```

```python
df_bronze.printSchema()
```

```python
bronze_path = "/Volumes/workspace/default/mis_datasets_mhealth/bronze"

(
    df_bronze.write
    .format("delta")
    .mode("overwrite")
    .save(bronze_path)
)
```

```python
bronze_df = spark.read.format("delta").load(bronze_path)

display(bronze_df)
```
